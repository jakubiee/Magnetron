```python
from magnetron.geometry import MagnetronGeometry
from magnetron.parameters import GeometryParameters, MeshSettings
from magnetron.mesh import MagnetronMesh, mesh_view
```

```python
magnetron_dimension = GeometryParameters(
    target_diameter=50.8,
    target_thickness=3,
    magnet_height=10,
    inner_magnet_diameter=18,
    outer_magnet_diameter=10,
    outer_magnet_ring_radius=22,
    magnet_to_target_distance=5.0,
    anode_cathode_gap=3.0,
    anode_width=4.0,
)

geometry = MagnetronGeometry(magnetron_dimension)
geometry.show()
```

```python
mesh_generator = MagnetronMesh(
    geometry=geometry, settings=MeshSettings(min_size=0.1, max_size=1.0)
)

mesh_path = mesh_generator.generate("../results/mesh.msh")

mesh_view(mesh_path)
```
